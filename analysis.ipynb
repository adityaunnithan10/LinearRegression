```python
import pandas as pd 
import matplotlib.pyplot as plt 
import main 
```

```python
data = pd.read_csv("data.csv")

plt.scatter(data.studytime, data.score, color="black")
plt.xlabel("Study time")
plt.ylabel("Score")
plt.title("Study time vs score")
plt.show()
```

```python
m, b = main.train(data, epochs=5000, L=0.00035)
```

```python
xs = [data.studytime.min(), data.studytime.max()]

plt.scatter(data.studytime, data.score, color="black", label="Data")
plt.plot(xs, [m * x + b for x in xs], color="red", label=f"y = {m:.3f}x + {b:.3f}")
plt.xlabel("Study time")
plt.ylabel("Score")
plt.legend()
plt.show()
```
